# __Preprocessing__

```python
# Create a Copy of the Text

df["clean_comments"] = df['Comments'].copy()
df.sample()
```

### __Lowercasing__

```python
#converting into lowercase

df['clean_comments'] = df['clean_comments'].str.lower()
df.sample()
```

### __Removing extra space__

```python
# Check spaces at the beginning or end
(df['clean_comments'] != df['clean_comments'].str.strip()).sum()
```

```python
# Removing the extra space

df['clean_comments'] = df['clean_comments'].str.strip()
```

```python
# verifying 
(df['clean_comments'] != df['clean_comments'].str.strip()).sum()
```

### __Punctuation__

```python
# cleaning punctuation in the text

import string

df['clean_comments'] = df['clean_comments'].str.translate(str.maketrans("","",string.punctuation))
df.sample()
```

### __Remove Extra Whitespace__

```python
#Removing extra whiespace in the text

import re
df['clean_comments'] = df['clean_comments'].str.replace(r'\s+'," ", regex=True).str.strip()
df.sample(5)
```

### __Normalize Repeated Characters__

```python
# Checking Repeation

df[df['clean_comments'].str.contains(r'(.)\1{2,}', regex=True, na=False)][
    ['Comments', 'clean_comments']
].head(5)
```

```python
# Cleaning it

import re

def normalize_repeated_chars(text):
    return re.sub(r'(.)\1{2,}', r'\1\1', text)

df['clean_comments'] = df['clean_comments'].apply(normalize_repeated_chars)
df.loc[360]
```

### __Stopwords__

```python
from collections import Counter

word_counts = Counter(
    word
    for comment in df['clean_comments']
    for word in comment.split()
)

word_counts.most_common(30)
```

```python
# Removing these stop words
stop_words = [
    'hai', 'ka', 'ki', 'ko', 'ho', 'ke', 'to',
    'se', 'ye', 'is', 'bhi', 'aur', 'kar',
    'hain', 'or', 'ha', 'me', 'mein',
    'tha', 'raha', 'ab'
]
```

```python
# Creating function for removing stop words
def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return ' '.join(filtered_words)
```

```python
df['clean_comments'] = df['clean_comments'].apply(remove_stopwords)
```

### __Spliting Data into features and targets__

```python
# Features
X = df['clean_comments']

# Target
y = df['Classes']
```

### __Spliting into training and testing data__

```python
from sklearn.model_selection import train_test_split

X_train, X_text, y_train, y_test = train_test_split(
    X,
    y,
    random_state=42,
    shuffle=True,
    stratify=y
)
```
