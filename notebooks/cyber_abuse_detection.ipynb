# __Data Cleaning__

```python
# Removing the trailing space

df['Classes'] = df['Classes'].str.strip()
```

```python
# Checking again

print(df['Classes'].unique())
print(df['Classes'].value_counts(dropna=False)) # dropna=False: it also count the NaN values
```

```python
# Na and Na are same(not abusive). Let's clear it make only one value

df['Classes'] = df['Classes'].replace('Na','NA')
```

```python
# Checking again

print(df['Classes'].unique())
print(df['Classes'].value_counts(dropna=False))
```
